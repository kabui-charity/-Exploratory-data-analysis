### data analysis

```python
#percentage of students placed in job
placement_data['status'].value_counts(normalize = True)* 100
```

```python
placement_data
```

```python
#does workex affect placement
pd.crosstab(placement_data['workex'], placement_data['status']) #cross tabulation
```

```python
#placement percentage for workex and status
placement_rate = pd.crosstab(placement_data['workex'], placement_data['status'], normalize = 'index')*100
placement_rate
```

```python
#visualizing workex and placement rate
placement_rate.plot(kind = 'bar', figsize = (6,4))
plt.title ('placement rate by work experience')
plt.ylabel('percentage')
plt.xlabel('workex')
plt.legend(title = 'placement')
plt.show()
```

```python
#does specialization affect placement
p_rate = pd.crosstab(placement_data['specialisation'], placement_data['status'], normalize = 'index') * 100
p_rate
```

```python
#visualizing specialisation and placement rate
p_rate.plot(kind = 'bar', figsize = (6,4))
plt.title ('placement rate by specialisation')
plt.ylabel('percentage')
plt.xlabel('specialisation')
plt.legend(title = 'placement')
plt.show()
```

```python
#gender comparison
place_rate =pd.crosstab(placement_data['status'], placement_data['gender'], normalize = 'index')* 100
place_rate
```

```python
#visualizing gender and placement rate
place_rate.plot(kind = 'bar', figsize = (6,4))
plt.title ('placement rate by gender')
plt.ylabel('percentage')
plt.xlabel('gender')
plt.legend(title = 'placement')
plt.show()
```
